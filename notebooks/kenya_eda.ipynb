# Kenya Climate Data Analysis (EDA)

This notebook analyzes Kenya climate data (2015–2026).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/kenya.csv")  
df["Country"] = "Kenya"
df.head()
```

```python
df["DATE"] = pd.to_datetime(df["YEAR"] * 1000 + df["DOY"], format="%Y%j")
df["MONTH"] = df["DATE"].dt.month
df.head()
```

## Data Cleaning

Replace missing values and remove duplicates.

```python
df = df.replace(-999, np.nan)
df = df.drop_duplicates()
df.isna().sum()
```

```python
df.describe()
```

```python
plt.figure()
plt.plot(df["DATE"], df["T2M"])
plt.title("Kenya Temperature Trend")
plt.xlabel("Date")
plt.ylabel("T2M")
plt.show()
```

```python
df.groupby("MONTH")["PRECTOTCORR"].mean().plot(kind="bar")
plt.title("Kenya Monthly Rainfall")
plt.show()
```

## Insights

- Temperature changes over time
- Rainfall shows seasonal patterns
